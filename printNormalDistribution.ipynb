# import all libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

## for normal dostribution 

```python

mu = 0      # mean
sigma = 1   # std-dev

```

```python
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# generate 1000 data point with equal space  between -4 to 4
```

```python
y = norm.pdf(x, mu, sigma)

#normal distribution f() assign  probobibity to all data point and far point less probobility and near point more probability  
```

```python
plt.plot(x, y, label=f'N({mu},{sigma**2})')
plt.xlabel('x')
plt.ylabel('Probability density')
plt.title('Normal distribution')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, y, label='Normal Distribution', color='blue')

# Plot mean line
plt.axvline(mu, color='black', linestyle='--', label='Mean (μ)')

# Plot standard deviation lines
for i in range(1, 4):
    plt.axvline(mu + i*sigma, color='red', linestyle='--', label=f'+{i}σ') #axvline is ued print all verical line
    plt.axvline(mu - i*sigma, color='red', linestyle=':', label=f'-{i}σ')

# Add labels and title
plt.title('Normal Distribution with Standard Deviation Lines')
plt.xlabel('X values')
plt.ylabel('Probability Density')
plt.legend(loc='upper left')

# Show the plot
plt.show()
```
